# Verification notebook

This notebook serves as a testing ground fro the implementation

```python
from pathlib import Path

from src.datasets import SemanticDroneDataset
from src.datasets.splits import split_sample_ids


PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

SEMANTIC_DRONE_ROOT = PROJECT_ROOT / "data" / "semantic_drone"

semantic_dataset = SemanticDroneDataset(SEMANTIC_DRONE_ROOT)

semantic_ids = [
    image_path.stem
    for image_path, _ in semantic_dataset.samples
]

print("Number of IDs:", len(semantic_ids))
print("First IDs:", semantic_ids[:5])
```

```python
split_result = split_sample_ids(semantic_ids, seed=420)

train_ids = split_result["train"]
val_ids = split_result["val"]
test_ids = split_result["test"]

assert len(train_ids) == 280
assert len(val_ids) == 60
assert len(test_ids) == 60

train_set = set(train_ids)
val_set = set(val_ids)
test_set = set(test_ids)
all_ids = set(semantic_ids)

assert train_set.isdisjoint(val_set)
assert train_set.isdisjoint(test_set)
assert val_set.isdisjoint(test_set)

assert train_set | val_set | test_set == all_ids

assert split_result == split_sample_ids(semantic_ids, seed=420)

print("Split verification passed, wohoo")
```
